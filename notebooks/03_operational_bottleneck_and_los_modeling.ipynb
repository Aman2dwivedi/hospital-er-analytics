### 3. Queueing Simulation: Peak Hours Physician Intervention

### 4. Visualizing Simulated Wait Time Savings

```python
plt.figure(figsize=(10, 5))
plt.plot(sim_df['additional_doctors_added'], sim_df['simulated_avg_wait_mins'], marker='o', color='crimson', linewidth=2.5)
plt.title('Simulated Peak Wait Time Reduction (ESI 3 Urgent Cohort)', fontsize=13, fontweight='bold')
plt.xlabel('Additional Physicians on Peak Shift')
plt.ylabel('Projected Wait Time (Minutes)')
plt.grid(True)
plt.show()
```
