```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style='whitegrid')
plt.rcParams['font.size'] = 11
plt.rcParams['figure.figsize'] = (12, 6)

# Load cleaned dataset
df = pd.read_csv(os.path.join('..', 'data', 'processed', 'cleaned_er_data.csv'))
df['arrival_timestamp'] = pd.to_datetime(df['arrival_timestamp'])
print(f"Loaded {len(df):,} patient records. Analysis by Aman Dwivedi.")
```

### 1. Hourly Patient Arrival Distribution & Peak Rush Hours

```python
plt.figure(figsize=(12, 5))
hourly_vol = df.groupby('arrival_hour')['patient_id'].count()
ax = sns.barplot(x=hourly_vol.index, y=hourly_vol.values, palette='Blues_r')
plt.title('Hourly Hospital ER Patient Arrivals (Diurnal Peaks at 11 AM & 7 PM)', fontsize=14, fontweight='bold')
plt.xlabel('Hour of Day (0 - 23)')
plt.ylabel('Patient Visits')
plt.show()
```

### 2. Door-to-Doctor Wait Time by ESI Triage Acuity

```python
plt.figure(figsize=(12, 6))
sns.boxplot(x='triage_level_name', y='wait_time_minutes', data=df, palette='Set2')
plt.title('Door-to-Doctor Wait Time Distribution across Triage Levels', fontsize=14, fontweight='bold')
plt.xlabel('ESI Triage Category')
plt.ylabel('Wait Time (Minutes)')
plt.show()
```

### 3. Correlation Matrix: Operational & Clinical Variables

```python
num_cols = ['age', 'wait_time_minutes', 'treatment_time_minutes', 'total_los_hours', 'bed_occupancy_rate_pct', 'total_cost_usd', 'patient_satisfaction_score']
corr = df[num_cols].corr()

plt.figure(figsize=(9, 7))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Feature Correlation Matrix (Wait Time vs Satisfaction r = -0.68)', fontsize=13, fontweight='bold')
plt.show()
```
